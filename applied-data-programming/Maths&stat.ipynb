```python
from scipy.optimize import root, root_scalar, brentq
from math import cos
def eqn(x): # f(x)
    return x + cos(x) # x is in radian
myroot = root(eqn,0)
print(myroot)
res_scalar = root_scalar(eqn, bracket=[-2,0])
print(res_scalar.root)
res_brentq = brentq(eqn, -2,0)
print(res_brentq)
```

## Hypothesis Testing (T-Test) 

```python
dat1 = np.random.normal(loc=10, scale=2, size=50) # mean 10 and 50 samples std 2
print(dat1)
dat2 = np.random.normal(loc=11, scale=2, size=50) # mean 10 and 50 samples
print(dat2)
```

```python
# Independat T-Test
t_stat, p_value = stat.ttest_ind(dat1,dat2)
print("T-stat:", t_stat)
print("P-value:", p_value)
```
