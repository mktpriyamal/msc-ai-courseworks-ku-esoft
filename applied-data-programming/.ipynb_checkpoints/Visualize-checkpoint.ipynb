# Introduction to Matplotlib in Python

Matploylib is module that contains classes and functions for creating MATLAB style graphics and plots with easy callable functions.

* Plots, histograms, power spectras, bar charts, scatterplots etc.
* http://matplotlib.org/gallery.html

**NumPy has to be available to use Matplotlib**

The primary sub module we use is 'pyplot' which is commonly reffered in  the programs as 'plt'

Let start by importing library module

```python
#!pip install matplotlib
import numpy as np
import matplotlib.pyplot as plt # Similarly -> from matplotlib import as plt
```

* Figure plot with axis object

```python
plt.plot([1,2,3,4,7,5]) # indexes are implicit
```

```python
# If the axes are needed to be highlighted, both axes can be given with the data
plt.plot([1,2,3,4,5,6],[1,4,6,8,9,4])
```

```python
# Change axes values
plt.plot([1,4,6,8,9,4],[1,2,3,4,5,6])
```

```python
# Define data variables and use them as input to plot
xVal = [1,2,3,4,5]
yVal = [1,4,6,4,9]
plt.plot(xVal,yVal)
print(type(xVal))
```

* Changing Default Plot Settings

```python
plt.plot(xVal,yVal, linewidth=5)
```

```python
# Colured changed ticks with shapes
plt.plot(xVal,yVal, 'r+')
```

```python
plt.plot(xVal,yVal, color='indigo', marker='o', markersize=12)
```

* Adding axes names and titles for the figure 

```python
plt.title('Sample Title Here', fontsize=18)
plt.xlabel('Label of X-axis', fontsize=14)
plt.ylabel('Label of Y-axis', fontsize=14)
plt.plot(xVal, yVal, marker='D', markersize=10)
plt.savefig('My_c_fig1.png') # save the figure in current folder 
plt.show()
```

```python
t = np.arange(0.0, 2, 0.01) # Time interval range
s1 = np.sin(2*np.pi*t) #sin values of each
s2 = np.cos(2*np.pi*t) #cos values of each
plt.figure(1)
plt.subplot(221)
plt.plot(t,s1)
plt.title('Sin Function')
plt.subplot(222)
plt.plot(t,s2)
plt.title('Cos Function')
plt.subplot(223)
plt.plot(s1,s2)
plt.title('Sin vs Cos')
plt.subplot(224)
plt.plot(t,s1+s2)
plt.title('Sin + Cos Function')
plt.tight_layout() # Adding space between the graphs
```

```python
fig1, ax = plt.subplots()
ax.plot(t,1+s1)
parameters = {'axes.labelsize':20, 'axes.titlesize':30}
plt.rcParams.update(parameters)
ax.set(xlabel='time (seconds)')
ax.set(ylabel='Voltage (mv)')
ax.set(title='Sine Function')
plt.tight_layout() # Adding space between the graphs
```

* Multiple plots within one Graph

```python
x1 = np.arange(1,100.5, 0.5) # Generate X axis
y1 = 2.0 * np.sqrt(x1)
y2 = 2.0 * x1**(1.0/3.0) # Cubic Root
plt.plot(x1,y1, label='y1') # Legend of y1
plt.plot(x1,y2, label='y2') # Legend of y2
plt.legend() # Ask to include the legends
plt.title('SQRT vs Cubic Root', fontsize=20)
plt.xlabel('X Parameter', fontsize=14)
plt.ylabel('Value Parameter', fontsize=14)
plt.tight_layout() # Adding space between the graphs
```

* Scatter Bubble Plot

```python
np.random.seed(19680801)
N=50
x3 = np.random.rand(N)
y3 = np.random.rand(N)
colors = np.random.rand(N)
area = (30*np.random.rand(N))**2 
plt.scatter(x3,y3, s=area, c=colors, alpha=0.5)
for i, txt in enumerate(np.round(area)):
    plt.annotate(txt,(x3[i],y3[i]))
```

* Bar Graph and Stack Bar Graph

```python
labels = ['G1', 'G2','G3','G4','G5']
men_mean = [20,35,30,37,27]
plt.bar(labels, men_mean)
plt.title('Mean Values')
```

```python
women_mean= [25,32,34,20,25]
men_std = [2,3,4,3,1]
women_std = [2,3,2,2,5]
width = 0.5 # Width of the bars
fig3, ax2 = plt.subplots()
ax2.bar(labels, men_mean, width, yerr=men_std, label='Men')
ax2.bar(labels, women_mean, width, yerr=women_std, bottom=men_mean, label='Women')
ax2.legend()
```

* Box Plot

```python
np.random.seed(19680801)
all_data = [np.random.normal(0,std,size=100) for std in range(1,4)]
labels1 = ['X1','X2','X3']
fig4, (ax4,ax5) = plt.subplots(nrows=1,ncols=2,figsize=(9,5))

# Rectangular Box Plot
bplot1 = ax4.boxplot(all_data, vert=True, # Vertical box plot alignment
                     patch_artist=True, #Fill with color
                     labels=labels1) # X axis labels
ax4.set_title('Rectangular Box Plot', size=15)

# notch shape Box Plot
bplot2 = ax5.boxplot(all_data, vert=True, # Vertical box plot alignment
                     notch=True,
                     patch_artist=True, #Fill with color
                     labels=labels1) # X axis labels
ax5.set_title('Notch Box Plot', size=15)

colors = ['pink', 'lightblue','lightgreen']

for bplot in (bplot1,bplot2):
    for patch, color in zip(bplot['boxes'], colors):
        patch.set_facecolor(color)

# Adding horizontal grid Lines
for ax in [ax4,ax5]:
    ax.yaxis.grid(True)
    ax.set_xlabel('Three seperate samples', size=12)
    ax.set_ylabel('Value',size=12)
```

* Multiple Bars

```python
# We use previous Men women data
x5 = np.arange(len(labels)) # locations for labels
width1 = 0.25 # Width of the bars
fig6, ax6 = plt.subplots()
rect1 = ax6.bar(x5 - width1/2, men_mean, width1, label='Men')
rect2 = ax6.bar(x5 + width1/2, women_mean, width1, label='Women')
ax6.set_title('Scores by group and gender', size=20)
ax6.set_ylabel('Scores', size=12)
ax6.set_xticks(x5)
ax6.set_xticklabels(labels)
ax6.legend()

ax6.bar_label(rect1, padding=0)
ax6.bar_label(rect2, padding=0)
fig6.tight_layout()
```

* Area Stack Plot

```python
# data from United Nations World Population Prospects (Revision 2019)
# https://population.un.org/wpp/, license: CC BY 3.0 IGO
year = [1950, 1960, 1970, 1980, 1990, 2000, 2010, 2018]
population_by_continent = {
    'africa': [228, 284, 365, 477, 631, 814, 1044, 1275],
    'americas': [340, 425, 519, 619, 727, 840, 943, 1006],
    'asia': [1394, 1686, 2120, 2625, 3202, 3714, 4169, 4560],
    'europe': [220, 253, 276, 295, 310, 303, 294, 293],
    'oceania': [12, 15, 19, 22, 26, 31, 36, 39]}
```

```python
fig7, ax7 = plt.subplots()
ax7.stackplot(year, population_by_continent.values(), 
              labels= population_by_continent.keys(),
              alpha=0.8)
ax7.legend(loc='upper left')
ax7.set_title('World Population (1950-2018)',size=20)
ax7.set_xlabel('Year', size=12)
ax7.set_ylabel('Number of People (Millions)', size=12)
fig7.savefig('population.png')
```
