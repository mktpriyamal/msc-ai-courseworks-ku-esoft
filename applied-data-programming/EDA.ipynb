## Core components of Pandas: Series and DataFrame
Pandas **Series** data structure is one dimensional data structure similar to an array whic can hold homogeneous data with index. There are several fundamental operations possible with this data structure.

Pandas **DataFrame** data structure can store data of different types (including character, int, float categories) in columns. It is similar to a spreadsheet, s SQL table or dataframe in R.

## Merge DataFrames

```python
df_left = pnd.DataFrame({'id':[1,2],
                         'Name':['Anil','Vishal'],
                         'subject_id':['sub1','sub2']})
print("Left DataFrame")
print(df_left)
df_right = pnd.DataFrame({'id':[1,2],
                          'Name':['Saman','Saliya'],
                          'subject_id':['sub2','sub4']})
print("Right DataFrame")
print(df_right)
```

```python
print("Merged DataFrames: based on ID:")
pnd.merge(df_left, df_right, on='id')
```

```python
print("Merged DataFrames: based on subject_id:")
pnd.merge(df_left, df_right, on='subject_id')
```

```python
print("Merged DataFrames: Right Join:")
pnd.merge(df_left, df_right, on='subject_id', how='right')
```

```python
print("Merged DataFrames: Left Join:")
pnd.merge(df_left, df_right, on='subject_id', how='left')
```

```python
print("Merged DataFrames: Outer Join:")
pnd.merge(df_left, df_right, on='subject_id', how='outer')
```

```python
print("Merged DataFrames: Inner Join:")
pnd.merge(df_left, df_right, on='subject_id', how='inner')
```

## Write a DataFrame to a File

```python
dfout = pnd.merge(df_left, df_right, on='subject_id', how='outer')
dfout.columns=['IDX','NAMEX','Subject','IDY','NAMEY']
dfout
```

```python
# Write to a csv data file
dfout.to_csv('out_df1.csv', index =False, header=True)
```

```python
# Write to a txt data file
dfout.to_csv('out_df1.txt', index =False, header=True)
```
