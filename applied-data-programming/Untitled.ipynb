```python
import numpy as np
```

```python
def angle_between(v1, v2, deg=True):
    v1, v2 = np.array(v1, dtype=float), np.array(v2, dtype=float)
    cos_theta = np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))
    angle = np.arccos(np.clip(cos_theta, -1.0, 1.0))
    return np.degrees(angle) if deg else angle
```

```python
u = [1, 0, 0]
v = [0, 1, 0]

angle_between(u, v)
```

```python
u = [1, 2, 3, 4, 5]
v = [6, 7, 8, 9, 0]

angle_between(u, v)
```
